```pyspark
from pyspark.sql import SparkSession

# Initialize Spark session
spark = SparkSession.builder \
    .appName("Yelp Reviews Loader") \
    .getOrCreate()
```

```pyspark
# Path to the JSON file on HDFS
hdfs_path = "/user/hadoop/output"

# Load the JSON data into a PySpark DataFrame
df = spark.read.json(hdfs_path)
```

```pyspark
# Show the schema to understand the structure of the DataFrame
df.printSchema()
```

```pyspark
# Show a sample of the data
df.count()
```

```pyspark
# Run a sample query to count the number of reviews per star rating
df.groupBy("stars").count().show()

# Run another sample query to filter reviews with at least 1 'useful' vote and display the top 10
df.filter(df.useful >= 1).select("review_id", "user_id", "business_id", "text", "useful").show(10, truncate=False)
```

```pyspark
df1 = df
df1 = df1.withColumnRenamed("business_id", "business_id_1")
df1 = df1.withColumnRenamed("cool", "cool_1")
df1 = df1.withColumnRenamed("date", "date_1")
df1 = df1.withColumnRenamed("funny", "funny_1")
df1 = df1.withColumnRenamed("review_id", "review_id_1")
df1 = df1.withColumnRenamed("stars", "stars_1")
df1 = df1.withColumnRenamed("text", "text_1")
df1 = df1.withColumnRenamed("useful", "useful_1")

df2 = df
df2 = df2.withColumnRenamed("business_id", "business_id_2")
df2 = df2.withColumnRenamed("cool",               "cool_2")
df2 = df2.withColumnRenamed("date",               "date_2")
df2 = df2.withColumnRenamed("funny", "funny_2")
df2 = df2.withColumnRenamed("review_id", "review_id_2")
df2 = df2.withColumnRenamed("stars", "stars_2")
df2 = df2.withColumnRenamed("text", "text_2")
df2 = df2.withColumnRenamed("useful", "useful_2")

# Merge (join) the DataFrames on 'user_id'
# Using 'inner' join to merge only the matching rows between df1 and df2
merged_df = df1.join(df2, on="user_id", how="inner")

# df1 = merged_df

# df3 = df
# df3 = df3.withColumnRenamed("business_id", "business_id_3")
# df3 = df3.withColumnRenamed("cool",               "cool_3")
# df3 = df3.withColumnRenamed("date",               "date_3")
# df3 = df3.withColumnRenamed("funny", "funny_3")
# df3 = df3.withColumnRenamed("review_id", "review_id_3")
# df3 = df3.withColumnRenamed("stars", "stars_3")
# df3 = df3.withColumnRenamed("text", "text_3")
# df3 = df3.withColumnRenamed("useful", "useful_3")

# # Merge (join) the DataFrames on 'user_id'
# # Using 'inner' join to merge only the matching rows between df1 and df2
# merged_df = df1.join(df3, on="user_id", how="inner")

merged_df.count()
```

```pyspark
# Create a temporary view so that you can run SQL queries on it
merged_df.createOrReplaceTempView("yelp_reviews")

# Run an SQL query to count the number of reviews per star rating
query_1 = spark.sql("""
    SELECT stars_1, COUNT(*) AS review_count 
    FROM yelp_reviews 
    GROUP BY stars_1
    ORDER BY stars_1
""")

# Show the result of the first query
#query_1.show()
```
