```pyspark
from pyspark.sql import SparkSession

# Initialize Spark session
spark = SparkSession.builder \
    .appName("Read CSV from HDFS") \
    .getOrCreate()

# Path to the file in HDFS
hdfs_path = "hdfs:///user/hadoop/employee_data.csv"

# Read the CSV file into a DataFrame
df = spark.read.csv(hdfs_path, header=True, inferSchema=True)

# Show the DataFrame contents
df.show()
```

```pyspark
# Step 1: Register the DataFrame as a temporary SQL view
df.createOrReplaceTempView("employees")

# Step 2: Use Spark SQL to calculate the average age
average_age_df = spark.sql("SELECT AVG(age) AS avg_age FROM employees")

# Step 3: Show the result
average_age_df.sh
```

```pyspark
average_age_df.write.csv("s3://emrdemomuf/output/hasil")
```
